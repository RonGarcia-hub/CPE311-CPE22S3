```python
#Exercise 1.1

import pandas as pd

df_aapl = pd.read_csv('aapl.csv')
df_amzn = pd.read_csv('amzn.csv')
df_fb = pd.read_csv('fb.csv')
df_goog = pd.read_csv('goog.csv')
df_nflx = pd.read_csv('nflx.csv')
```

```python
#Exercise 1.2

df_aapl['Ticker'] = "AAPL"
df_amzn['Ticker'] = "AMZN"
df_fb['Ticker'] = "FB"
df_goog['Ticker'] = "GOOG"
df_nflx['Ticker'] = "NFLX"
```

```python
#Exercise 1.3

dfs_list = [df_aapl, df_amzn, df_fb, df_goog, df_nflx]
df_faang = pd.concat(dfs_list)
```

```python
#Exercise 1.4

df_faang.to_csv('faang.csv', index=False)
```

```python
#Exercise 2.1

from datetime import date, datetime, time


df_faang["date"] = pd.to_datetime(df_faang["date"])


df_faang["volume"] = df_faang["volume"].astype("int64")


df_faang = df_faang.sort_values(by=["date", "Ticker"])

df_faang.head(11)
```
