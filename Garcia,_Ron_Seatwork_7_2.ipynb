```python
import pandas as pd

df = pd.read_csv('Meteorite_Landings.csv')
df.head(11)
```

##limiting the years

```python
df['year'] = pd.to_datetime(df['year'], errors='coerce')
df['year'] = df['year'].dt.year

df_filtered = df[(df['year'] >= 2005) & (df['year'] <= 2009)]

df.head()
```

```python
df_filtered['mass (g)'] = pd.to_numeric(df_filtered['mass (g)'], errors='coerce')

pivot = df_filtered.pivot_table(
    index='year',
    columns='fall',  # 'Found' vs 'Fell'
    values='mass (g)',
    aggfunc=['count', lambda x: x.quantile(0.95)]
    )

pivot
```
