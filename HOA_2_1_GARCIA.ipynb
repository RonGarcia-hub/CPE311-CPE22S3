```python
Rings = 4
Source = 'Start'
Auxiliary = 'Middle'
Destination = 'Goal'

pegs = {
    'Start': list(range(Rings, 0, -1)),
    'Middle': [],
    'Goal': []
}

def print_state():
    print("\nPeg states:")
    for peg_name, disks in pegs.items():
        if disks: print(f"{peg_name}: [{ ' | '.join(map(str,disks)) }]")
        else: print(f"{peg_name}: []")
    print("-" * 25)

def state(from_peg, to_peg):
    disk = pegs[from_peg].pop(0)
    pegs[to_peg].insert(0, disk)


def TowerofHanoi(Rings, Source, Auxiliary, Destination):
  if Rings == 1:
    print ("Move Disk 1 from", Source, "to", Destination)
    state(Source, Destination)
    print_state()
    return

  #Stage 1
  TowerofHanoi(Rings - 1, Source, Destination, Auxiliary)
  print ("Move Disk", Rings, "from", Source, "to", Destination)
  state(Source, Destination)
  print_state()

  #Stage 2
  TowerofHanoi(Rings - 1, Auxiliary, Source, Destination)

TowerofHanoi(Rings, Source, Auxiliary, Destination)
```

In this actvity I used recursion to solve the tower of hanoi dividing it into two stages wherein hte stages had different eng goals, stage 1 had the middle rod as the end goal which is called auxilliary in the beggign but the end goal for stage 1. Stage 2 is where the code switches the end doal to the original end goal which was called destination and where auxilliary is now the source.
